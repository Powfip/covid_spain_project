 # Notebook 04: Cargar datos a la base de datos

- Tenemos configurada las rutas dinamicas
- EL .env configurado
- Tabla creada en MySQL
> Procedemos a volcar los datos a la tabla

```
import pandas as pd
import sys
from pathlib import Path
from sqlalchemy import text

# Añadimos las rutas
sys.path.append(str(Path.cwd().parent))
from src.config import PROCESSED_DATA_DIR
from src.database_mgr import get_engine

# Cargamos los datos
df_daily = pd.read_csv(PROCESSED_DATA_DIR / "daily_stats_spain.csv")
df_master = pd.read_csv(PROCESSED_DATA_DIR / "master_df.csv")

# Conectamos a la basa de datos
engine = get_engine()
# Subir los datos
try:
    print("🚀 Subiendo datos a MySQL...")
    df_daily.to_sql("daily_stats", con=engine, if_exists="replace", index=False)
    df_master.to_sql("master_data", con=engine, if_exists="replace", index=False)
    print("✅ ¡Carga completada con éxito!")
except Exception as e:
    print(f"❌ Algo falló: {e}")
```

```
with engine.connect() as conn:
    # Contamos cuántas filas hay en la tabla maestra
    res = conn.execute(text("SELECT COUNT(*) FROM master_data"))
    print(f"La base de datos tiene {res.fetchone()[0]} registros listos para el dashboard.")
```
